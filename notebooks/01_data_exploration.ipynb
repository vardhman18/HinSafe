# Data Exploration

```python
import pandas as pd

pd.set_option('display.max_colwidth', 100)

print("Libraries loaded successfully and pandas display option set.")
```

```python
HATECOMMENTS_PATH = "../datasets/raw/hatecomments.csv"
CYBERBULLYING_PATH = "../datasets/raw/cyberbullying_dataset.csv"
HINGLISH_PATH = "../datasets/raw/hinglish.csv"

print("File paths are set ")
```

```python
def explore_dataset(df, name, label_column, text_column=None):
    """Display basic information about a dataset."""

    separator = "=" * 55

    print(f"\n{separator}")
    print(f"DATASET: {name}")
    print(separator)

    print(f"\nShape: {df.shape[0]} rows, {df.shape[1]} columns")
    print(f"Columns: {df.columns.tolist()}")

    print("\nFirst five rows:")
    display(df.head())

    print("\nMissing values:")
    print(df.isnull().sum())

    if label_column in df.columns:
        print(f"\nLabel distribution ({label_column}):")
        print(df[label_column].value_counts())
    else:
        print(f"\nWarning: Column '{label_column}' was not found.")

    duplicate_rows = df.duplicated().sum()
    print(f"\nDuplicate rows: {duplicate_rows}")

    if text_column and text_column in df.columns:
        duplicate_text = df.duplicated(subset=[text_column]).sum()
        print(f"Duplicate text entries: {duplicate_text}")
    elif text_column:
        print(f"Warning: Text column '{text_column}' was not found.")

    if duplicate_rows > 0:
        print("\nRecommendation: Remove duplicate rows during preprocessing.")
    else:
        print("\nNo duplicate rows found.")

    print(f"\n{separator}\n")
```

```python
# HATECOMMENTS dataset
try:
    df_hatecomments = pd.read_csv(HATECOMMENTS_PATH)
    print("HATECOMMENTS dataset loaded")
except FileNotFoundError:
    print(f"File not found at: {HATECOMMENTS_PATH}")
    print("Check the path in Section 1 and try again.")
```

```python
explore_dataset(df_hatecomments, name="HATECOMMENTS", label_column="label", text_column="text")
```

```python
#  Cyberbullying dataset
try:
    df_cyberbullying = pd.read_csv(CYBERBULLYING_PATH)
    print("Cyberbullying dataset loaded ")
except FileNotFoundError:
    print(f"File not found at: {CYBERBULLYING_PATH}")
    print("Check the path in Section 1 and try again.")
```

```python
explore_dataset(df_cyberbullying, name="Hinglish Cyberbullying", label_column="Label", text_column="Text")
```

```python
# Hinglish dataset
try:
    df_hinglish = pd.read_csv(HINGLISH_PATH)
    print("Hinglish dataset loaded")
except FileNotFoundError:
    print(f"File not found at: {HINGLISH_PATH}")
    print("Check the path in Section 1 and try again.")
```

```python
explore_dataset(df_hinglish, name="Hinglish Hate Speech", label_column="hate_label", text_column="text")
```

```python
summary = pd.DataFrame({
    "Dataset": ["HATECOMMENTS", "Hinglish Cyberbullying", "Hinglish Hate Speech"],
    "Rows": [df_hatecomments.shape[0], df_cyberbullying.shape[0], df_hinglish.shape[0]],
    "Columns": [df_hatecomments.shape[1], df_cyberbullying.shape[1], df_hinglish.shape[1]],
    "Label Column": ["label", "Label", "hate_label"],
})


total_rows = summary["Rows"].sum()

print("Dataset Summary")
display(summary)
print(f"\nTotal comments across all 3 datasets: {total_rows}")
```
